```python
import pandas as pd
df = pd.read_csv('Crop_Yield_Prediction.csv')
op = df['Yield']
df.drop(columns=['Yield'], inplace=True)
```

```python
# One-hot encode categorical features
df_encoded = pd.get_dummies(df, drop_first=True)
df_encoded = df_encoded.astype(int)

print(df_encoded.head())

# label_encoder = LabelEncoder()

# # Apply LabelEncoder to each feature in the list
# df['Crop'] = label_encoder.fit_transform(df['Crop'])
# df_encoded = df
# print(df_encoded.head())
# correlation_matrix = df_encoded.corr()
# plt.figure(figsize=(50, 40))
# sns.heatmap(correlation_matrix, cmap='BrBG', annot=True, fmt=".2f", linewidths=.5)
# plt.show()
```

```python
nan_dfc = [feature for feature in df.columns if df[feature].isnull().sum()>1 and df[feature].dtype=='O']
for f in nan_dfc:
    print("{} : {}% nans {}".format(f, np.round(df[f].isnull().mean(), 4), df[f].dtype))
```

```python
scrl = StandardScaler()
df_scaled = scrl.fit_transform(df_encoded)  # Scale the features

# Split data
x_train, x_test, y_train, y_test = train_test_split(df_scaled, op, test_size=0.2, random_state=42)

# Create and train the model
model = RandomForestRegressor()
model.fit(x_train, y_train)
model.score(x_test, y_test)
```

# Crop recomnd

```python
df = pd.read_csv('Crop_recommendation.csv')

op = df.label
df.drop(columns=['label'], inplace=True)
df.head()
```

```python
label_encoder = LabelEncoder()

# Apply LabelEncoder to each feature in the list
y = label_encoder.fit_transform(op)

# Verify the encoding
print(y)
```

```python
x = df
model = RandomForestClassifier()
model.fit(x, y)

# New data for prediction
new_data = [[75, 45, 40, 25, 80.5, 7, 250]]
new_data_df = pd.DataFrame(new_data, columns=x.columns)  # Use same columns as training data

# Make predictions
predictions = model.predict(new_data_df)

# Assuming label_encoder is already fitted
decoded_predictions = label_encoder.inverse_transform(predictions)

print(decoded_predictions)
```
